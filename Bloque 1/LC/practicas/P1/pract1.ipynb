```python
string = 'El/DT perro/N come/V carne/N de/P la/DT carnicería/N y/C de/P la/DT nevera/N y/C canta/V el/DT la/N la/N la/N ./Fp'

string = string.lower()

def insert_key_count(dictionary, key):
    if dictionary.get(key):
        dictionary[key] += 1
    else:
        dictionary[key] = 1
    return dictionary
```

```python

cat_dict = {}
for pair in string.split(' '):
    word,cat = pair.split('/')
    insert_key_count(cat_dict,cat)

print(cat_dict)
```

```python
word_span_cat_dict = {}
for pair in string.split(' '):
    word,cat = pair.split('/')
    if word_span_cat_dict.get(word):
        word_info = word_span_cat_dict[word]
        word_info[0] = word_info[0]+1
        insert_key_count(word_info[1], cat)
    else:
        word_info = [1, {cat:1}]
    word_span_cat_dict[word] = word_info
    

print(word_span_cat_dict)
```

```python
string_copy = '<S> ' + string + ' </S>'
string_copy_split = string_copy.split(' ')
pair_span_dict = {}
window = 2
for i in range(len(string_copy_split)-window):
    tup = []
    for item in string_copy_split[i:i+window]:
        tup.append( item.upper() if '<' in item else item.split('/')[1].upper() )
    tup = tuple(tup)
    insert_key_count(pair_span_dict, tup)
    
print(pair_span_dict)
```

```python
def get_prob(w):
    if not word_span_cat_dict.get(w):
        print('word not found')
    else:
        categories = word_span_cat_dict.get(w)[1].keys()
        counts = word_span_cat_dict.get(w)[1].values()
    for i in range(len(categories)):
        prob_lex = list(counts)[i]/sum(list(counts))
        prob_em = word_span_cat_dict[w][1][list(categories)[i]]/cat_dict[list(categories)[i]]
        print('P( {} | {} ) = {}'.format(list(categories)[i],w,prob_lex))
        print('P( {} | {} ) = {}'.format(w,list(categories)[i],prob_em))
        
get_prob('la')

```
