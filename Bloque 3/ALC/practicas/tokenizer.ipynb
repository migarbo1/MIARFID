```python
import re
```

```python
digit_group_re = re.compile(r'(\d+[,\.]\d+)')
digit_test = '45.54'
print(re.match(digit_group_re, digit_test))
```

```python
date_group_re = re.compile(r'(\d{1,2}[/\-]\d{1,2}[/\-]\d{2,4})')
date_test_1 = '12-03-23'
date_test_2 = '12/03/2021'
print(re.match(date_group_re, date_test_1))
print(re.match(date_group_re, date_test_2))
```

```python

hour_group_re = re.compile(r'(\d{1,2}:\d{1,2})')
hour_test_1 = '09:30'
hour_test_2 = '9:10'
print(re.match(hour_group_re, hour_test_1))
print(re.match(hour_group_re, hour_test_2))
```

```python
written_date_group_re = re.compile(r'(\d{1,2}\s+de\s+[A-Za-z]+(\s+de\s+\d{4}){0,1})')
written_date_test_1 = '12 de enero de 2013'
written_date_test_2 = '31 de agosto'
print(re.match(written_date_group_re, written_date_test_1))
print(re.match(written_date_group_re, written_date_test_2))
```

```python
url_group_re = re.compile(r'(http(s){0,1}://.*\.[a-z]{2,4})')
url_test_1 = 'https://www.upv.es'
url_test_2 = 'http://www.crazygames.com'
print(re.match(url_group_re, url_test_1))
print(re.match(url_group_re, url_test_2))
```

```python
email_group_re = re.compile(r'.*@.*\.[a-z]{2,4}')
email_test_1 = 'migarbo1@inf.upv.es'
email_test_2 = 'migarbo1.90.sabata@hotmail.com'
print(re.match(email_group_re, email_test_1))
print(re.match(email_group_re, email_test_2))
```

```python
mention_group_re = re.compile(r'(@[a-zA-Z0-9\_\-\.]+)')
mention_test_1 = '@migarbo1'
mention_test_2 = '@m.i.g.a.r.b.o'
mention_test_3 = '@migar_bo1'
mention_test_4 = '@migar-bo1'
print(re.match(mention_group_re, mention_test_1))
print(re.match(mention_group_re, mention_test_2))
print(re.match(mention_group_re, mention_test_3))
print(re.match(mention_group_re, mention_test_4))
```

```python
hashtag_group_re = re.compile(r'(#[a-zA-Z0-9\_]+)')
hashtag_test_1 = '#MeEncanta'
hashtag_test_2 = '#in_creible'
hashtag_test_3 = '#whatA_practica'
print(re.match(hashtag_group_re, hashtag_test_1))
print(re.match(hashtag_group_re, hashtag_test_2))
print(re.match(hashtag_group_re, hashtag_test_3))
```

```python
acronym_group_re = re.compile(r'(([A-Z]{1,2}\.)+)')
acronym_test_1 = 'U.R.R.S.'
acronym_test_2 = 'U.S.A.'
acronym_test_3 = 'EE.UU.'
print(re.match(acronym_group_re, acronym_test_1))
print(re.match(acronym_group_re, acronym_test_2))
print(re.match(acronym_group_re, acronym_test_3))
```

```python
emoji_group_re = re.compile(r'([\u263a-\U0001f645])')
emoji_test_2 = '🎉'
print(re.match(emoji_group_re, emoji_test_2))
```

```python
name_group_re = re.compile(r'(([A-ZÀ-ÿ][a-zÀ-ÿ]+\s+){3})')
name_test_1 = 'Miguel García Bohigues '
name_test_2 = 'Sofia Madagascar Pelota '
print(re.match(name_group_re, name_test_1))
print(re.match(name_group_re, name_test_2))
```

```python
treatment_group_re = re.compile(r'[SD][a-zªº]{1,4}\.')
treatment_test_1 = 'Dr.'
treatment_test_2 = 'Dra.'
treatment_test_3 = 'Sra.'
treatment_test_4 = 'Sª.'
print(re.match(treatment_group_re, treatment_test_1))
print(re.match(treatment_group_re, treatment_test_2))
print(re.match(treatment_group_re, treatment_test_3))
print(re.match(treatment_group_re, treatment_test_4))
```

```python
splitter_regex = re.compile(r'(?<! )(?=[.,!?():;])|(?<=[.,!?():;])(?! )')
point_test = 'ppp.'
point_test = 'ppp;'
print(re.sub(splitter_regex, '\n', point_test))
```

```python
ends_with_pointer_re = re.compile(r'.*[.,!?():;]')
```

```python
def tokenizer(sentence):
    out = ''
    regex_groups = [digit_group_re, date_group_re, hour_group_re, written_date_group_re, url_group_re, email_group_re, mention_group_re, hashtag_group_re, acronym_group_re, emoji_group_re, name_group_re, treatment_group_re]
    #print('sentence: {}'.format(sentence))
    #print('sentence_split: {}'.format(sentence.split(' ')))
    for word in sentence.split(' '):
        match = False
        if word == '':
            continue
        for regex in regex_groups:
            #if the current word is an exception, add it as it is
            if re.match(regex, word) != None:
                if re.match(ends_with_pointer_re,word):
                    word = word[:-1] + '\n'+word[-1]
                out += '{}\n'.format(word)
                match = True
            #if not, split the punctuation symbol and the word
        if not match:
            substitution = re.sub(splitter_regex, r'\n', word)
            substitution = substitution[:-1] if substitution.endswith('\n') else substitution
            if substitution != '':
                out += '{}\n'.format(substitution)
    return out
```

```python
def get_all_matches(input):
    matches = []
    for regex in [written_date_group_re, name_group_re]:
        matches += [m for m in re.finditer(regex, input)]
    return sorted(matches, key=lambda m: m.start())
```

```python
def pre_tokenizer(input):
    out = '## {}\n'.format(input)
    start = 0
    if re.search(written_date_group_re, input) or re.search(name_group_re, input):
        for match in get_all_matches(input):
            out += tokenizer(input[start:match.start()])
            out += '{}\n'.format(input[match.start(): match.end()])
            start = match.end()
        out += tokenizer(input[start:])

    else:
        out += tokenizer(input)

    return out

def tokenize(input):
    return pre_tokenizer(input)
```

```python
print(tokenize(input))
```
