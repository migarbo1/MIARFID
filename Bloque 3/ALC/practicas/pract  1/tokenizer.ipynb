```python
import re
```

```python
digit_group_re = r'(\d+[,\.]\d+)'
digit_test = '45.54'
print(re.match(digit_group_re, digit_test))
```

```python
date_group_re = r'(\d{1,2}[/\-]\d{1,2}[/\-]\d{2,4})'
date_test_1 = '12-03-23'
date_test_2 = '12/03/2021'
print(re.match(date_group_re, date_test_1))
print(re.match(date_group_re, date_test_2))
```

```python

hour_group_re = r'(\d{1,2}:\d{1,2})'
hour_test_1 = '09:30'
hour_test_2 = '9:10'
print(re.match(hour_group_re, hour_test_1))
print(re.match(hour_group_re, hour_test_2))
```

```python
written_full_date_group_re = r'(\d{1,2})\s+(de)\s+(enero|febrero|marzo|abril|mayo|junio|julio|agosto|septiembre|octubre|noviembre|diciembre)\s+(de)\s+(\d{4})'
written_date_group_re = r'(\d{1,2})\s+(de)\s+(enero|febrero|marzo|abril|mayo|junio|julio|agosto|septiembre|octubre|noviembre|diciembre)'

written_date_test_1 = '12 de enero de 2013'
written_date_test_2 = '31 de agosto'
print(re.match(written_date_group_re, written_date_test_1))
print(re.match(written_date_group_re, written_date_test_2))

print(re.sub(written_full_date_group_re, r'\1_\2_\3_\4_\5', written_date_test_1))
print(re.sub(written_date_group_re, r'\1_\2_\3', written_date_test_2))
```

```python
url_group_re = r'(http(s){0,1}://(\w|7|~|\-|/)+\.((\w|7|~|\-|/)+\.{0,1})+)'
url_test_1 = 'https://www.upv.es'
url_test_2 = 'http://www.crazygames.com'
url_test_3 = 'http://users.dsic.upv.es/~fpla/'
print(re.match(url_group_re, url_test_1))
print(re.match(url_group_re, url_test_2))
print(re.match(url_group_re, url_test_3))
```

```python
email_group_re = r'([\w\_\.\-]+@\w+(\.[a-z]{2,4})+)'
email_test_1 = 'migarbo1@inf.upv.es'
email_test_2 = 'migarbo1.90.sabata@hotmail.com'
print(re.match(email_group_re, email_test_1))
print(re.match(email_group_re, email_test_2))
```

```python
mention_group_re = r'(@[a-zA-Z0-9\_\-]+)'
mention_test_1 = '@migarbo1'
mention_test_2 = '@m.i.g.a.r.b.o'
mention_test_3 = '@migar_bo1'
mention_test_4 = '@migar-bo1'
print(re.match(mention_group_re, mention_test_1))
print(re.match(mention_group_re, mention_test_2))
print(re.match(mention_group_re, mention_test_3))
print(re.match(mention_group_re, mention_test_4))
```

```python
hashtag_group_re = r'(#[a-zA-Z0-9\_]+)'
hashtag_test_1 = '#MeEncanta'
hashtag_test_2 = '#in_creible'
hashtag_test_3 = '#whatA_practica'
print(re.match(hashtag_group_re, hashtag_test_1))
print(re.match(hashtag_group_re, hashtag_test_2))
print(re.match(hashtag_group_re, hashtag_test_3))
```

```python
acronym_group_re = r'(([A-Z]{1,2}\.)+)'
acronym_test_1 = 'U.R.R.S.'
acronym_test_2 = 'S.A.'
acronym_test_3 = 'EE.UU.:'
print(re.match(re.compile(acronym_group_re), acronym_test_1))
print(re.match(re.compile(acronym_group_re), acronym_test_2))
print(re.match(re.compile(acronym_group_re), acronym_test_3))
re.sub(acronym_group_re, r'\1 ', acronym_test_3)
```

```python
emoji_group_re = r'([\U0001F1E0-\U0001F1FF]|[\U0001F300-\U0001F5FF]|[\U0001F600-\U0001F64F]|[\U0001F680-\U0001F6FF]|[\U0001F700-\U0001F77F]|[\U0001F780-\U0001F7FF]|[\U0001F800-\U0001F8FF]|[\U0001F900-\U0001F9FF]|[\U0001FA00-\U0001FA6F]|[\U0001FA70-\U0001FAFF]|[\U00002702-\U000027B0]|[\U000024C2-\U0001F251])'
emoji_test_2 = '🎉'
print(re.match(emoji_group_re, emoji_test_2))
```

```python
name_group_re = r'([A-ZÀ-ÿ][a-zÀ-ÿ]+)\s+([A-ZÀ-ÿ][a-zÀ-ÿ]+)\s+([A-ZÀ-ÿ][a-zÀ-ÿ]+)\s*'
name_test_1 = 'Miguel García Bohigues '
name_test_2 = 'Sofia Madagascar Pelota '
print(re.match(name_group_re, name_test_1))
print(re.match(name_group_re, name_test_2))
print(re.sub(name_group_re, r'\1_\2_\3', name_test_1))
```

```python
treatment_group_re = r'([SD][a-zªº]{,4}\.)'
treatment_test_1 = 'Dr.'
treatment_test_2 = 'Dra.'
treatment_test_3 = 'Sra.'
treatment_test_4 = 'Sª.'
print(re.match(treatment_group_re, treatment_test_1))
print(re.match(treatment_group_re, treatment_test_2))
print(re.match(treatment_group_re, treatment_test_3))
print(re.match(treatment_group_re, treatment_test_4))
```

```python
fraction_group_regex = r'(\d+[\\/]\d+)'
fraction_test_1 = '2/3'
fraction_test_2 = '23/32'
print(re.match(fraction_group_regex, fraction_test_1))
print(re.match(fraction_group_regex, fraction_test_2))
```

```python
filename = './entrada_tokenizador_2023.txt'
output = ''
regex = '|'.join([written_full_date_group_re, written_date_group_re, name_group_re, date_group_re, url_group_re, email_group_re, digit_group_re, hour_group_re, mention_group_re, hashtag_group_re, emoji_group_re, acronym_group_re, treatment_group_re, fraction_group_regex, r'(\.\.\.|\.|,|;|:|!|¡|\?|¿|%|\'|")', r'([\w|\-]+)'])
with open(filename, 'r') as file:
    for line in file.readlines():
        output += '## {}'.format(line)
        all_matches = re.findall('({})'.format(regex), line)
        parsed = '\n'.join([b[0] for b in all_matches])
        output += parsed +'\n'

with open('output.txt', 'w') as file:
    file.write(output)
```
