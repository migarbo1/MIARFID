```python
prob_emp = [1/4, 1/4, 1/4, 1/4]

def f1(x,y):
    return int(x.__contains__('a') and y==0)

def f2(x,y):
    return int(x.__contains__('b') and y==0)

def f3(x,y):
    return int(x.__contains__('a') and y==1)

def f4(x,y):
    return int(x.__contains__('b') and y==1)

exp_val_func = [1/4, 1/4, 1/2, 1/2]
func = [f1, f2, f3, f4]
lamda = [0,0,0,0]
train_samples = [('aa', 0), ('bb', 0), ('ab', 1), ('ba', 1)]
```

```python
import numpy as np  
import math
def get_cond_prob(x,y):
    num = den = 0
    for i in range(4):
        loc_val = lamda[i] * func[i](x,y)
        num += loc_val
    for j in range(2):
        aux = 0
        for k in range(4):
            loc_val = lamda[k] * func[k](x,j)
            aux += loc_val
        den += math.exp(aux)
    
    res = math.exp(num)/den
    return res

def get_p_lambda_fi(index):
    res = 0
    for pair in train_samples:
        local_res = 1/4 * get_cond_prob(*pair) * func[index](*pair)
        res += local_res
    return res

def get_delta_i(fun_ind):
    p_lambda_fi = get_p_lambda_fi(fun_ind)
    num = exp_val_func[fun_ind]/p_lambda_fi
    print('p(fi): {}'.format(exp_val_func[fun_ind]))
    print('p_l(fi) = {}'.format(p_lambda_fi))
    print('p(fi)/p_l(fi) = {}'.format(num))
    print('log2(p(fi)/p_l(fi))= {}'.format(np.log(num)))
    return np.log(num) * (1 if fun_ind < 2 else 1/2)
```

```python
lamda = [lamda[i] + get_delta_i(i) for i in range(4)]
```
